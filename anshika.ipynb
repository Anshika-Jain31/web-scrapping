```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

all_data = []
page = 1

while True:
    url = f"https://npino.com/lookup/207ri0200x-infectious-disease/?page={page}"
    print(f"🔄 Scraping page {page}...")

    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'
    }

    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        print(f"❌ Failed to fetch page {page}. Status code: {response.status_code}")
        break

    soup = BeautifulSoup(response.text, 'html.parser')
    doctors = soup.find_all("div", class_="panel-body")

    if not doctors:
        print("✅ No more doctors found. Scraping complete.")
        break

    for block in doctors:
        try:
            # Doctor name
            name_tag = block.find("a", class_="question")
            name = name_tag.text.strip() if name_tag else "N/A"

            # Specialty
            specialty_tag = block.find("small")
            specialty = specialty_tag.text.strip().split("\n")[0].strip() if specialty_tag else "N/A"

            # NPI number
            npi_links = block.find_all("a", class_="question")
            npi = npi_links[1].text.strip() if len(npi_links) > 1 else "N/A"

            # Address, Phone, Fax
            address, phone, fax = "N/A", "N/A", "N/A"
            small_tags = block.find_all("small")
            if small_tags:
                for tag in small_tags:
                    text = tag.text
                    if "Address:" in text:
                        if "Phone:" in text:
                            address = text.split("Address:")[1].split("Phone:")[0].strip()
                        if "Phone:" in text:
                            phone = text.split("Phone:")[1].split("Fax:")[0].strip()
                        if "Fax:" in text:
                            fax = text.split("Fax:")[1].strip()

            all_data.append({
                "Name": name,
                "Specialty": specialty,
                "NPI": npi,
                "Address": address,
                "Phone": phone,
                "Fax": fax
            })
        except Exception as e:
            print(f"⚠ Error parsing a block: {e}")
            continue

    page += 1
    time.sleep(0.5)

# Save to Excel
df = pd.DataFrame(all_data)
df.to_excel("all_infectiousdisease_npino.xlsx", index=False)
print("✅ Data saved to 'all_infectious_disease_npino.xlsx'")
```
